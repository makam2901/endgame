## Discretization Tasks
3. **Optimal Binning Strategy Analysis**  
   Use KBinsDiscretizer to transform 'LotArea' into 5 bins using uniform, quantile, and k-means strategies. Evaluate which approach maximizes correlation with log(SalePrice).

4. **Temporal Feature Engineering**  
   Discretize 'YearBuilt' into decade-based categories and assess impact on linear model performance compared to raw numeric treatment

```python
df['YearBuilt'].unique()
```

```python
min_year = df['YearBuilt'].min()
max_year = df['YearBuilt'].max()

bins = np.arange((min_year//10)*10, ((max_year+1)//10)*10 + 1, 10)
labels = [f'decade of {x}' for x in bins[:-1]]
```

```python
df['YearBuiltEncoded'] = pd.cut(df['YearBuilt'], bins=bins, labels=labels)
```

```python
df['YearBuiltEncoded']
```

## Outlier Detection Tasks
13. **IQR-Based Filtering**  
    Identify outliers in 'GrLivArea' using 1.5*IQR rule and analyze their impact on sale price prediction error[7][18].

```python
Q1 = df['GrLivArea'].quantile(0.25)
Q3 = df['GrLivArea'].quantile(0.75)

IQR_15 = 1.5*(Q3-Q1)
outliers = df[(df['GrLivArea']< Q1 - IQR_15) | (df['GrLivArea']>Q3 + IQR_15)]
outliers
```

14. **Z-Score Anomaly Detection**  
    Flag homes with Z-scores >3 in 'TotalBsmtSF' and evaluate their characteristics compared to the overall population[7][12].

```python
mean_bsmt = df['TotalBsmtSF'].mean(skipna=True)
std_bsmt = df['TotalBsmtSF'].std(skipna=True, ddof=1)

Z_scores = np.abs(df['TotalBsmtSF'] - mean_bsmt)/(std_bsmt)

df[Z_scores > 3]
```
