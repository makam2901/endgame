### Task 4: Aggregation

**Description**: Compute summary statistics from the data.

**Formula 1 Exercise**:
```python
# Calculate the total number of wins for each constructor (team)
# Find the average finishing position for each driver across all seasons
```

````python
results[results['position']=='1'][['constructorId']].merge(constructors[['constructorId', 'name']], how='left', on='constructorId')[['name']].value_counts().sort_values(ascending=False)
````

````python
results['position'] = results['position'].apply(pd.to_numeric, errors='coerce')
results.dropna(inplace=True)
driver_avg_positions = (
    results[['driverId', 'position']]
    .merge(drivers[['driverId', 'forename', 'surname']], how='left', on='driverId')
    [['forename', 'surname', 'position']]
    .groupby(['forename', 'surname'])
    .mean()
    .reset_index()
)
driver_avg_positions.sort_values(by='position', ascending=False)
````

## Sample Solution Outline

Here's a sample outline for the Formula 1 joining task:

```python
import pandas as pd

# Load the required datasets
races = pd.read_csv('races.csv')
results = pd.read_csv('results.csv')
drivers = pd.read_csv('drivers.csv')
constructors = pd.read_csv('constructors.csv')

# Filter for 2020 season
races_2020 = races[races['year'] == 2020]

# Join the datasets
merged_results = results.merge(races_2020, on='raceId')
merged_results = merged_results.merge(drivers, on='driverId')
merged_results = merged_results.merge(constructors, on='constructorId')

# Select and rename columns for clarity
final_results = merged_results[[
    'name_x', 'date', 'forename', 'surname', 'nationality_x', 
    'name_y', 'position', 'points'
]].rename(columns={
    'name_x': 'race_name',
    'forename': 'driver_first_name',
    'surname': 'driver_last_name',
    'nationality_x': 'driver_nationality',
    'name_y': 'team_name'
})

# Sort by race date and position
final_results = final_results.sort_values(['date', 'position'])

# Display the results
print(final_results.head(20))
```
